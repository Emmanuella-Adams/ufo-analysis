# Now we detect outliers

```python
Q1 =np.percentile(valid_duration , 25)
Q3 = np.percentile(valid_duration , 75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
outliers = valid_duration[
    (valid_duration < lower) |
    (valid_duration > upper)
]
```

```python
len(outliers)
```

```python
len(outliers) /len(valid_duration) * 100
```

## We now use normalisation to normalise the valid duration

```python
normalized = (
    valid_duration - np.min(valid_duration) 
    ) / (
        np.max(valid_duration) - np.min(valid_duration)
    )
```
