# UFO Sightings Visualization

## Objective

In this notebook, I explore the cleaned UFO sightings dataset through visualizations. The goal is to uncover trends, distributions, and relationships between variables before using the data for quantum computing experiments.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
clean_df =pd.read_csv('../data/processed/ufo_clean.csv')
```

## Top 10 UFO shapes

```python
shape_counts = (
    clean_df['UFO_shape']
    .value_counts()
    .head(10)
)

plt.figure(figsize = (10 , 5))
```

```python
plt.bar(shape_counts.index , shape_counts.values)
plt.title("Top 10 UFO Shapes")
plt.xlabel("Shape")
plt.ylabel("Number of Sightings")
plt.show()
```

```python
country_counts = clean_df["country"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(country_counts.index, country_counts.values)
plt.title("Sightings by Country")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()
```

```python
duration = clean_df["length_of_encounter_seconds"].dropna()

plt.figure(figsize=(8,5))
plt.hist(duration, bins=50)
plt.title("Distribution of Encounter Duration")
plt.xlabel("Seconds")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    clean_df["longitude"],
    clean_df["latitude"],
    s=2,
    alpha=0.5
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Geographical Distribution of UFO Sightings")
plt.show()
```

```python
clean_df["Date_time"] = pd.to_datetime(clean_df["Date_time"])

year_counts = clean_df["Date_time"].dt.year.value_counts().sort_index()

plt.figure(figsize=(10,5))
plt.plot(year_counts.index, year_counts.values)
plt.title("UFO Sightings Over Time")
plt.xlabel("Year")
plt.ylabel("sightings")
plt.show()
```

```python
cities = clean_df['city'].value_counts().head(15)
plt.figure(figsize = (20, 6))
plt.bar(cities.index , cities.values)
plt.title("Top 15 Cities with Ufo Sightings")
plt.xlabel("Sightings")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(duration)
plt.title("Encounter Duration")
plt.ylabel("Seconds")
plt.show()
```

# Reflection
 - most sightings occur usually less than a second
 - the encounter duration is highly skewed
 - the boxplot reveals many extreme outliers
 - the geograpgic scatterplot shows concentration in the US.
